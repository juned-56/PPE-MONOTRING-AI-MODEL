```python
import os
root = dataset.location
for split in ["train", "valid", "test"]:
    image_dir = os.path.join(root, split, "images")
    label_dir = os.path.join(root, split, "labels")
    if os.path.exists(image_dir):
        images = [
            f for f in os.listdir(image_dir)
            if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))
        ]
        labels = [
            f for f in os.listdir(label_dir)
            if f.endswith(".txt")
        ]
        print(f"\n{split.upper()}")
        print("Images :", len(images))
        print("Labels :", len(labels))
```

```python
from ultralytics import YOLO
model = YOLO("/content/yolov12s.pt")
print("YOLOv12s loaded successfully")
```

```python
results = model.train(
    data="/content/PPE-FULL-5/data.yaml",
    epochs=100,
    imgsz=640,
    batch=8,
    device=0,
    workers=2,
    patience=20,
    project="/content/PPE_TRAINING",
    name="ppe_yolov12s",
    exist_ok=True
)
```
